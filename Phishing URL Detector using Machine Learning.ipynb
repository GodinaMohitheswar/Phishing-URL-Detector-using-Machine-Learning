```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

# Load the dataset
df = pd.read_csv('/content/dataset_phishing.csv')

# Data Preprocessing
df = df.dropna()  # Drop missing values
df['status'] = df['status'].apply(lambda x: 1 if x == 'phishing' else 0)

# Feature Selection
selected_features = [
    "length_url", "length_hostname", "nb_dots", "nb_hyphens",
    "nb_at", "nb_qm", "nb_and", "nb_or", "ip", "web_traffic",
    "google_index", "page_rank"
]

X = df[selected_features]
y = df['status']

# Model Training
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = RandomForestClassifier()
model.fit(X_train, y_train)

# Model Evaluation
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Feature Extraction for URL input
def extract_features(url):
    return [
        len(url),
        len(url.split('/')[2].split(':')[0]),  # Extract length of hostname - THIS WAS MISSING
        url.count('.'),
        url.count('-'),
        url.count('@'),
        url.count('?'),
        url.count('&'),
        url.count('|'),
        1 if any(char.isdigit() for char in url.split('/')[0]) else 0,
        0,  # Placeholder for web traffic
        0,  # Placeholder for Google index
        0   # Placeholder for PageRank
    ]
# Interactive Prediction
while True:
    url = input("\nEnter a URL to check (or type 'exit' to quit): ")
    if url.lower() == 'exit':
        print("Exiting the Phishing URL Detector. Goodbye!")
        break

    features = extract_features(url)
    prediction = model.predict([features])[0]

    if prediction == 1:
        print(f"🔴 The URL '{url}' is classified as Phishing.")
    else:
        print(f"🟢 The URL '{url}' is classified as Legitimate.")
```
